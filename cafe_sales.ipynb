```python
# Create a copy of your dataset
df_copy = df.copy()

df_copy.info()
```

## Column standardization

```python
# Standardazing column names - remove extra spaces, before and after, lower case the column namw and replace spaces with underscores

df_copy.columns
df_copy.columns = df_copy.columns.str.strip().str.lower().str.replace(' ', '_', regex=False)

df_copy.info()
```

### Removing duplicates

```python
# first check for duplicates
df_copy.duplicated().sum() # returns the sum of duplicates in the dataset
```

```python
# drop duplicates
df_copy.drop_duplicates(inplace=True)

df_copy.info()
```

### Correcting data types

```python
df_copy.head(10)
```

```python
# convert numeric columns - quantity, price_per_unit and total_spent, to numeric datatype

for col in ['quantity', 'price_per_unit', 'total_spent']:
    df_copy[col] = pd.to_numeric(df_copy[col], errors='coerce')
```

```python
df_copy.info()
```

### Handling missing Values

```python
# have a quick review of count of missing values per column
df_copy.isnull().sum()
```

```python
# imputate the missing values with a specific value or statistical measures like mean, median or mode of the column

# impute quantity and price_per unit column missing valuse using median
quantity_median = df_copy['quantity'].median()
price_median = df_copy['price_per_unit'].median()

print(quantity_median, price_median)

df_copy['quantity'].fillna(quantity_median, inplace=True)
df_copy['price_per_unit'].fillna(price_median, inplace=True)

# impute total_spent column missing values using quantity column value multiplied by price_per_unit column value
df_copy['total_spent'].fillna(df_copy['quantity'] * df_copy['price_per_unit'], inplace=True)

# verify if there is any missing values in the imputed columns
df_copy.isnull().sum()
```
